```python
import heapq

def prim(graph, start=0):
    visited = set()
    min_heap = [(0, start, -1)]  # (weight, current_vertex, parent)
    mst = []
    total_cost = 0

    while min_heap:
        weight, vertex, parent = heapq.heappop(min_heap)

        if vertex in visited:
            continue

        visited.add(vertex)

        # Add edge to MST (except for starting vertex)
        if parent != -1:
            mst.append((parent, vertex, weight))
            total_cost += weight

        # Add neighboring edges to heap
        for neighbor, edge_weight in graph[vertex]:
            if neighbor not in visited:
                heapq.heappush(
                    min_heap,
                    (edge_weight, neighbor, vertex)
                )

    return mst, total_cost


# Example graph
graph = {
    0: [(1, 2), (3, 6)],
    1: [(0, 2), (2, 3), (3, 8), (4, 5)],
    2: [(1, 3), (4, 7)],
    3: [(0, 6), (1, 8), (4, 9)],
    4: [(1, 5), (2, 7), (3, 9)]
}

mst, cost = prim(graph)

print("Edges in Minimum Spanning Tree:")
for u, v, weight in mst:
    print(f"{u} -- {v} = {weight}")

print("Total cost:", cost)
```
